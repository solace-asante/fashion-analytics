```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

path = '../data/processed/merged_cleaned_2019.csv'
```

```python
color_col = ['product_group_name','colour_group_name','price']

df = pd.read_csv(path,usecols=color_col)

df.head()
```

```python

colour_counts = df['colour_group_name'].value_counts()
```

```python
dist = pd.DataFrame({
    'purchases': colour_counts,
    'share': colour_counts / colour_counts.sum()
})
dist['cumulative_share'] = dist['share'].cumsum()

print(len(dist), 'distinct colours')
dist.round(4)
```

```python

top_colours = dist.head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_colours['share'], y=top_colours.index, color='steelblue')
plt.title('Top 10 colours by share of purchases')

plt.xlabel('Share of all purchases')
plt.ylabel('Colour')

plt.savefig('../reports/top_colours.png', dpi=150, bbox_inches='tight')

plt.show()
```

## Findings

The data has 50 colour names. Black is the most common, with 34.6% of purchases. It takes 13 colours to reach 80% of all purchases, so a small number of colours account for most of what is bought.
