```python
import pandas as pd
```

```python
fashion_df = pd.read_csv('data/processed/hm_2019.csv')

print(fashion_df.shape)
print(fashion_df.columns.tolist())
print(fashion_df.dtypes)
```

```python
print(fashion_df.isnull().sum())
```

```python
print(fashion_df.duplicated().sum())
```

```python
duplicates =fashion_df[fashion_df.duplicated(keep=False)]
duplicates.sort_values(by=['customer_id', 'article_id', 't_dat']).head(30) # Duplicate rows are kept because they represent genuine repeat purchases of the same item, per H&M's official dataset documentation.
```

```python
print(fashion_df['age'].mean())
print(fashion_df['age'].median())
print(fashion_df['age'].max())
print(fashion_df['age'].min())

```

```python
fashion_df['age']=fashion_df['age'].fillna(fashion_df['age'].median())
```

```python
fashion_df['club_member_status'] =fashion_df['club_member_status'].fillna('Unknown')
```

```python
print(fashion_df.isnull().sum())
```

```python
fashion_df['t_dat'] = pd.to_datetime(fashion_df['t_dat'])
```

```python
fashion_df['age'] = fashion_df['age'].astype(int)
```

```python
print(fashion_df.dtypes)
```

```python
fashion_df.to_csv('data/processed/merged_cleaned_2019.csv', index=False)
```
